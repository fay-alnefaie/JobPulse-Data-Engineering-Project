```python
import pandas as pd
from datetime import datetime
```

```python
# Loading the data
data = pd.read_csv('jadarat_dataset.csv')
data.head()
```

```python
# Checking the columns of this dataset
data.columns
```

```python
# Checking the details of the dataset
data.info()
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
```

```python
# Checking the missing values
data.isnull().sum()
```

```python
# Checking numerical columns
data.select_dtypes(include='number')
```

```python
# Checking categorical columns
data.select_dtypes(include='object')
```

```python
# Checking the duplicated rows in the dataset
print("Number of duplicate rows:", data.duplicated().sum())
```

```python

d = data[data.duplicated(
    subset=["job_title", "comp_name", "city","positions"],
    keep=False
)]

print(d)
```

```python
df = data.copy()
df.columns = df.columns.str.strip().str.lower()
print(df.columns)
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
print("Duplicates after text cleaning:", df.duplicated().sum())
```

```python
df = df.drop_duplicates(keep="first")
print("Duplicates after removal:", df.duplicated().sum())
```

```python
# converting date type to date instead of object
df["job_date"] = pd.to_datetime(
    df["job_date"],
    errors="coerce")
print(df["job_date"].dtype)
```

```python
# checking the conversion of the dates
print("Invalid dates:", df["job_date"].isna().sum())
print("Earliest date:", df["job_date"].min())
print("Latest date:", df["job_date"].max())
```

```python
print(df["positions"].unique())
print(df[df["positions"] <= 0])
```

```python
# checking for negative values
print(df[df["exper"] < 0])
```

```python
print("Final rows:", len(df))
print("Final columns:", len(df.columns))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Select specific job titles from job_title column

print(df["job_title"].value_counts())
```

```python
df.columns
```

```python
# Removing extra columns
df = df.drop(columns=['comp_type','comp_size' ,'benefits',
                 'contract','positions','gender'])
print(df.columns)
```

```python
# UNIFY COLUMN NAMES with other datasets
df = df.rename(columns={
    'comp_name': 'company_name',
    'eco_activity': 'industry',
    'contract': 'contract_type',
    'exper': 'experience_years',
    'job_date':'job_posted_date'

})

print(df.columns)
```

```python
# Adding new columns to unify the dataset with other datasets

df[['url', 'company_url',
    'employment_type','job_category','description' ]] = None


df['source_job_id'] = range(1, len(df) + 1)
df['country'] = 'Saudi Arabia'
df['job_source'] = 'Jadarat.com'
df['workplace_type'] = 'Onsite'
df['is_salary_disclosed'] = df['salary'].notna()
df['job_status'] = 'Closed'
df['collected_at'] = datetime.now().strftime('%Y-%m-%d %H:%M:%S')

df.info()
```

```python
# Removing salary column after comsidering "is_salary_discloed" column

df.drop(columns=['salary'], inplace=True)
```

```python
# Checking new updates
df.info()
```

```python
# Rearrange the columns
df = df[
    [
        'source_job_id',
        'job_title',
        'url',
        'company_name',
        'company_url',
        'city',
        'region',
        'country',
        'employment_type',
        'workplace_type',
        'experience_years',
        'job_category',
        'industry',
        'is_salary_disclosed',
        'job_source',
        'job_posted_date',
        'job_status',
        'collected_at',
        'description'
    ]
]
```

```python
# Save the cleaned dataset
df.to_csv("jadarat_cleaned.csv", index=False)
```
